```python
import math

documents = [
    "I love natural language processing",
    "Natural language processing is interesting",
    "I love machine learning"
]

# Convert documents into words
docs = []

for d in documents:
    docs.append(d.lower().split())

# Create vocabulary
vocabulary = []

for doc in docs:
    for word in doc:
        if word not in vocabulary:
            vocabulary.append(word)

# Calculate TF-IDF
for i in range(len(docs)):

    print("\nDocument", i + 1)

    for word in vocabulary:

        # TF
        tf = docs[i].count(word) / len(docs[i])

        # Count documents containing the word
        df = 0

        for doc in docs:
            if word in doc:
                df = df + 1

        # IDF
        idf = math.log(len(docs) / df)

        # TF-IDF
        tfidf = tf * idf

        if tf > 0:
            print(word,
                  "\nTF =", round(tf, 3),
                  "\nIDF =", round(idf, 3),
                  "\nTF-IDF =", round(tfidf, 3))
```
