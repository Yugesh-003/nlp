## Bigram own

```python
text = "I like mango I like apple"

# Convert sentence into words
words = text.split()

# Create bigrams
bigrams = []

for i in range(len(words) - 1):
    bigram = (words[i], words[i + 1])
    bigrams.append(bigram)

print("Words:", words)
print("Bigrams:", bigrams)
```

```python
text = "I like mango I like apple banana orange I like "

words = text.split()

for i in range(len(words) - 1):

    first_word = words[i]
    second_word = words[i + 1]

    first_word_count = 0
    bigram_count = 0

    # Count first word and bigram
    for j in range(len(words) - 1):

        if words[j] == first_word:
            first_word_count += 1

        if words[j] == first_word and words[j + 1] == second_word:
            bigram_count += 1

    # Calculate probability
    probability = bigram_count / first_word_count

    print("Bigram:", first_word, second_word)
    print("Count:", bigram_count, "/", first_word_count)
    print("Probability:", probability)
    print()
```
